```python
import pandas as pd

df = pd.read_csv("churn-bigml-20.csv")
```

```python
df.columns = (
    df.columns
    .str.strip()
    .str.lower()
    .str.replace(" ", "_")
)
```

```python
df["international_plan_flag"] = df["international_plan"].map({"Yes":1,"No":0})
df["voice_mail_plan_flag"] = df["voice_mail_plan"].map({"Yes":1,"No":0})
```

```python
df["total_traffic_minutes"] = (
    df["total_day_minutes"]
    + df["total_eve_minutes"]
    + df["total_night_minutes"]
    + df["total_intl_minutes"]
)

df["total_calls"] = (
    df["total_day_calls"]
    + df["total_eve_calls"]
    + df["total_night_calls"]
    + df["total_intl_calls"]
)

df["total_cost"] = (
    df["total_day_charge"]
    + df["total_eve_charge"]
    + df["total_night_charge"]
    + df["total_intl_charge"]
)

df["avg_minutes_per_call"] = df["total_traffic_minutes"] / df["total_calls"]
```

```python
df["capacity_risk"] = pd.cut(
    df["total_traffic_minutes"],
    bins=[0,550,700,10000],
    labels=["Normal","Warning","Critical"]
)

df["incident_level"] = pd.cut(
    df["customer_service_calls"],
    bins=[-1,1,3,100],
    labels=["Low","Medium","High"]
)

df["cost_per_minute"] = df["total_cost"] / df["total_traffic_minutes"]
df["calls_per_minute"] = df["total_calls"] / df["total_traffic_minutes"]
```

```python
df.to_csv("infra_capacity_engineered.csv", index=False)
```
